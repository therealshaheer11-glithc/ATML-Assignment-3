```python
# Task 3 — Block 2D. Hardware accelerator: None.
import os, hashlib, shutil, subprocess, threading, json
from pathlib import Path
from google.colab import runtime

accelerator = os.environ.get("COLAB_GPU", "0").strip().lower() not in ("", "0", "false", "none")
accelerator |= any(os.environ.get(k) for k in ("COLAB_TPU_ADDR", "TPU_NAME", "TPU_WORKER_ID"))
accelerator |= os.environ.get("PJRT_DEVICE", "").upper() == "TPU"
accelerator |= bool(list(Path("/dev").glob("nvidia[0-9]*")))
if shutil.which("nvidia-smi"):
    try:
        probe = subprocess.run(
            ["nvidia-smi", "--query-gpu=name", "--format=csv,noheader"],
            capture_output=True, text=True, timeout=10
        )
        accelerator |= probe.returncode != 0 or bool(probe.stdout.strip())
    except (OSError, subprocess.TimeoutExpired):
        accelerator = True

if accelerator:
    print("CPU-only task. Requesting runtime release.", flush=True)
    def release_accelerator():
        try:
            runtime.unassign()
        except Exception:
            print("Use Runtime > Disconnect and delete runtime.", flush=True)
    thread = threading.Thread(target=release_accelerator, daemon=True)
    thread.start()
    thread.join(20)
    raise RuntimeError("Reconnect with accelerator None before rerunning.")

import pyarrow.parquet as pq

root = Path("/content/drive/MyDrive/ATML_PA3_Task3")
if not root.is_dir():
    raise RuntimeError("Existing Drive workspace unavailable. Stop and share output.")

for folder in ("qwen25-math", "qwen25vl-mathvista"):
    relative = folder + "/generations.parquet"
    path = root / "sources/supplied" / relative
    receipt_path = (
        root / "manifests/source_files"
        / (relative.replace("/", "__") + ".json")
    )
    receipt = json.loads(receipt_path.read_text())

    h = hashlib.sha256()
    with path.open("rb") as stream:
        for chunk in iter(lambda: stream.read(4 * 1024 * 1024), b""):
            h.update(chunk)
    if h.hexdigest() != receipt["sha256"]:
        raise RuntimeError("Source differs from saved receipt: " + relative)

    table = pq.ParquetFile(path)
    identities = set()
    for batch in table.iter_batches(
        batch_size=512, columns=["model", "dataset"]
    ):
        for row in batch.to_pylist():
            identities.add((row["model"], row["dataset"]))

    print(json.dumps({
        "folder": folder,
        "rows": table.metadata.num_rows,
        "identifiers": [
            {"model": model, "dataset": dataset}
            for model, dataset in sorted(identities)
        ],
        "verified_source_receipt": receipt
    }, indent=2), flush=True)

print("\nSUPPLIED README\n", flush=True)
print((root / "sources/supplied/README.MD").read_text(), flush=True)
print("CPU DIAGNOSTIC COMPLETE — keep accelerator None.", flush=True)
```

```python
# Block 16: CPU-only recovery snapshot.
# No installs, generation, or Drive writes.
import hashlib, json, os, shutil, subprocess, threading, uuid, zipfile
from datetime import datetime, timezone
from pathlib import Path
from google.colab import drive, files, runtime

def require(condition, message):
    if not condition:
        raise RuntimeError(message)

# Refuse an accelerator allocation before mounting Drive.
evidence = []
if os.environ.get("COLAB_GPU", "0").strip().lower() not in (
    "", "0", "false", "none"
):
    evidence.append("GPU allocation variable")
if (
    any(os.environ.get(k) for k in (
        "COLAB_TPU_ADDR", "TPU_NAME", "TPU_WORKER_ID"
    ))
    or os.environ.get("PJRT_DEVICE", "").upper() == "TPU"
):
    evidence.append("TPU allocation variable")
if list(Path("/dev").glob("nvidia[0-9]*")):
    evidence.append("GPU device")
if shutil.which("nvidia-smi"):
    try:
        probe = subprocess.run(
            ["nvidia-smi", "--query-gpu=name", "--format=csv,noheader"],
            capture_output=True, text=True, timeout=10
        )
        if probe.returncode != 0 or probe.stdout.strip():
            evidence.append("GPU present or allocation uncertain")
    except (OSError, subprocess.TimeoutExpired):
        evidence.append("Allocation probe unavailable")

if evidence:
    def release():
        try:
            runtime.unassign()
        except Exception:
            pass

    task = threading.Thread(target=release, daemon=True)
    task.start()
    task.join(10)
    raise RuntimeError(
        "Select accelerator None before running Block 16. "
        "Release requested; if connected, disconnect and delete runtime. "
        + ", ".join(evidence)
    )

print("CPU-only recovery snapshot. No installation or model execution.",
      flush=True)
drive.mount("/content/drive", timeout_ms=120000)
root = Path("/content/drive/MyDrive/ATML_PA3_Task3")
require(root.is_dir() and not root.is_symlink(),
        "Task 3 workspace unavailable")
work = root / "results/task3_live_generation_A002"
require((work / "calls").is_dir(),
        "A002 saved-call folder unavailable")

def inventory():
    paths = list(work.rglob("*"))
    status = root / "live/session_status"
    if status.exists():
        paths.extend(status.rglob("*"))
    paths.extend((root / "logs").glob("gpu_*.json"))

    for relative in (
        "live/task3_live_inputs_v1.json",
        "configs/MANUAL_REQUIREMENTS.json",
        "configs/task3_live_generation_recipe_A002.json",
        "configs/task3_live_GPU_environment_A002.json",
        "configs/task3_live_runtime_preflight.json",
        "provenance/task3_gpu_code_A002.zip",
    ):
        path = root / relative
        require(path.is_file(), "Required evidence missing: " + relative)
        paths.append(path)

    require(not any(p.is_symlink() for p in paths),
            "Unexpected symlink; preserve for review")
    return sorted(set(p for p in paths if p.is_file()))

paths = inventory()
require(len(paths) <= 5000,
        "Unexpectedly many evidence files; preserve for review")
sizes = {p: p.stat().st_size for p in paths}
require(
    all(n <= 32 * 1024**2 for n in sizes.values())
    and sum(sizes.values()) <= 150 * 1024**2,
    "Unexpected evidence size; preserve for review"
)

blobs = {p: p.read_bytes() for p in paths}
require(paths == inventory(),
        "Evidence inventory changed; stop and share output")
require(
    all(len(blobs[p]) == sizes[p] and p.read_bytes() == blobs[p]
        for p in paths),
    "Evidence changed during snapshot; stop and share output"
)

manifest = {
    "schema_version": 1,
    "created_UTC": datetime.now(timezone.utc).isoformat(),
    "CPU_only": True,
    "model_executed": False,
    "Drive_files_modified": False,
    "scope": (
        "Mounted-filesystem snapshot; full call validity "
        "and interruption cause require review."
    ),
    "files": [
        {
            "path": p.relative_to(root).as_posix(),
            "bytes": len(raw),
            "sha256": hashlib.sha256(raw).hexdigest(),
        }
        for p, raw in blobs.items()
    ],
}

output = Path("/content") / (
    "Task_3_Block_16_CPU_Recovery_" + uuid.uuid4().hex[:8] + ".zip"
)
with zipfile.ZipFile(
    output, "x", compression=zipfile.ZIP_DEFLATED
) as archive:
    for p, raw in blobs.items():
        archive.writestr(p.relative_to(root).as_posix(), raw)
    archive.writestr(
        "SNAPSHOT_MANIFEST.json",
        json.dumps(manifest, indent=2) + "\n"
    )

with zipfile.ZipFile(output) as archive:
    require(archive.testzip() is None,
            "Local snapshot archive failed verification")
    require(
        all(archive.read(p.relative_to(root).as_posix()) == raw
            for p, raw in blobs.items()),
        "Archive readback differs"
    )

canonical = [
    p for p in paths
    if p.parent == work / "calls" and p.suffix == ".json"
]
pending = [
    p.name for p in paths
    if p.parent == work / "calls" and ".pending_" in p.name
]
print(json.dumps({
    "snapshot": output.name,
    "files": len(blobs),
    "canonical_call_files": len(canonical),
    "pending_call_files": pending,
    "archive_bytes": output.stat().st_size,
    "archive_sha256": hashlib.sha256(output.read_bytes()).hexdigest(),
    "note": (
        "File count only; saved-call validity and stop reason "
        "not yet independently checked."
    ),
}, indent=2), flush=True)

print("Keep accelerator None. Send the downloaded ZIP and this output.",
      flush=True)
files.download(str(output))
```
